```python
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from google.colab import files
from IPython.display import display

```

```python
up = files.upload()
nama_file = list(up.keys())[0]
print("File terbaca:", nama_file)
```

```python
JUMLAH_KOLOM_DEMOGRAFI = 5  # A-E: Timestamp, Kelas, Jenis Kelamin, Jumlah Platform, Jumlah Premium

df = pd.read_excel(nama_file)

print("Jumlah responden :", len(df))
print("Jumlah kolom     :", df.shape[1])
```

```python
kolom_likert = [
    c for c in df.columns[JUMLAH_KOLOM_DEMOGRAFI:]
    if "[" in str(c)
]

butir = df[kolom_likert].copy()

print("Jumlah butir Likert :", butir.shape[1])
print("Data kosong         :", int(butir.isna().sum().sum()))
```

```python
skala = {
    "Sangat Tidak Setuju (STJ)": 1,
    "Sangat Tidak Setuju (STS)": 1,  # cadangan jika ada variasi label
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5,
}
```

```python
butir_clean = butir.apply(lambda col: col.astype(str).str.strip())

num = butir_clean.replace(skala).apply(pd.to_numeric, errors="coerce")

gagal = int(num.isna().sum().sum())
print("Sel yang gagal dikonversi:", gagal)

if gagal > 0:
    print("\nKolom yang masih mengandung nilai tidak terbaca:")
    print(num.isna().sum()[num.isna().sum() > 0])
    raise ValueError(
        "Ada jawaban yang tidak cocok dengan kamus skala. "
        "Periksa label jawaban pada data."
    )

display(num.head())
```

```python
dimensi = {}

for kolom in num.columns:
    match = re.match(r"^(.+?)\s*\[", str(kolom))
    if match is None:
        print("Kolom tidak cocok format dimensi:", kolom)
        continue

    nama_dimensi = match.group(1).strip()
    dimensi.setdefault(nama_dimensi, []).append(kolom)

print("\nDimensi dan jumlah butir:")
for nama, daftar in dimensi.items():
    print(f"- {nama}: {len(daftar)} butir")
```

```python
n = len(num)
t = stats.t.ppf(0.975, n - 2)
r_tabel = t / np.sqrt((n - 2) + t**2)

print(f"\nn = {n} | df = {n-2} | r tabel (α=0,05; dua sisi) = {r_tabel:.3f}")

hasil_validitas = []

for nama, daftar in dimensi.items():
    x = num[daftar]
    total = x.sum(axis=1)

    for i, kolom in enumerate(daftar, 1):
        r, p = stats.pearsonr(x[kolom], total)
        r_koreksi, _ = stats.pearsonr(x[kolom], total - x[kolom])

        hasil_validitas.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "r hitung": round(r, 3),
            "Corrected r": round(r_koreksi, 3),
            "p-value": round(p, 4),
            "Keterangan": "Valid" if r > r_tabel else "Tidak valid",
        })

tabel_validitas = pd.DataFrame(hasil_validitas)

print("\n=== TABEL VALIDITAS ===")
display(tabel_validitas)
```

```python
def cronbach_alpha(x):
    x = x.dropna()
    k = x.shape[1]
    if k < 2:
        return np.nan
    return k / (k - 1) * (
        1 - x.var(ddof=1).sum() / x.sum(axis=1).var(ddof=1)
    )

def kategori_alpha(a):
    if pd.isna(a):
        return "Tidak dapat dihitung"
    if a >= 0.90:
        return "Sangat baik"
    elif a >= 0.80:
        return "Baik"
    elif a >= 0.70:
        return "Reliabel"
    else:
        return "Tidak reliabel"

hasil_reliabilitas = []
alpha_dihapus = []

for nama, daftar in dimensi.items():
    x = num[daftar]
    a = cronbach_alpha(x)

    hasil_reliabilitas.append({
        "Dimensi": nama,
        "Jumlah butir": len(daftar),
        "Cronbach's alpha": round(a, 3),
        "Kategori": kategori_alpha(a),
    })

    # Alpha jika butir dihapus, hanya untuk dimensi > 2 butir
    if len(daftar) > 2:
        for i, kolom in enumerate(daftar, 1):
            a_del = cronbach_alpha(x.drop(columns=kolom))
            alpha_dihapus.append({
                "Dimensi": nama,
                "Butir": f"B{i}",
                "Alpha jika dihapus": round(a_del, 3),
            })

tabel_reliabilitas = pd.DataFrame(hasil_reliabilitas)
tabel_alpha_dihapus = pd.DataFrame(alpha_dihapus)

print("\n=== TABEL RELIABILITAS ===")
display(tabel_reliabilitas)

print("\n=== ALPHA JIKA BUTIR DIHAPUS ===")
display(tabel_alpha_dihapus)
```

```python
gabung = tabel_validitas.merge(tabel_alpha_dihapus, on=["Dimensi", "Butir"], how="left")

with pd.ExcelWriter("hasil_uji_validitas_reliabilitas.xlsx") as w:
    gabung.to_excel(w, sheet_name="Validitas", index=False)
    tabel_reliabilitas.to_excel(w, sheet_name="Reliabilitas", index=False)

files.download("hasil_uji_validitas_reliabilitas.xlsx")
```

```python

import matplotlib.pyplot as plt
import numpy as np

# Visualisasi uji validitas setiap butir
jumlah_dimensi = len(dimensi)

fig, axes = plt.subplots(
    nrows=jumlah_dimensi,
    ncols=1,
    figsize=(11, max(3.2, jumlah_dimensi * 2.7)),
    squeeze=False
)

axes = axes.flatten()

for ax, (nama_dimensi, daftar) in zip(axes, dimensi.items()):
    data_plot = tabel_validitas[
        tabel_validitas["Dimensi"] == nama_dimensi
    ].copy()

    nilai = data_plot["r hitung"].astype(float).to_numpy()
    label_butir = data_plot["Butir"].tolist()

    # Warna kuning untuk valid dan merah untuk tidak valid
    warna = [
        "#F2C94C" if keterangan == "Valid" else "#D83B3B"
        for keterangan in data_plot["Keterangan"]
    ]

    posisi = np.arange(len(data_plot))

    ax.barh(
        posisi,
        nilai,
        color=warna,
        height=0.62,
        zorder=3
    )

    ax.set_yticks(posisi)
    ax.set_yticklabels(label_butir, fontsize=10)
    ax.invert_yaxis()

    # Garis batas nilai r tabel
    ax.axvline(
        x=r_tabel,
        color="#555555",
        linestyle="--",
        linewidth=1.4,
        zorder=4
    )

    # Mengatur rentang sumbu X
    nilai_min = min(0, float(np.min(nilai)) if len(nilai) else 0)
    nilai_max = max(
        r_tabel,
        float(np.max(nilai)) if len(nilai) else r_tabel
    )

    rentang = max(nilai_max - nilai_min, 0.2)

    ax.set_xlim(
        nilai_min - rentang * 0.22,
        nilai_max + rentang * 0.30
    )

    # Label nilai untuk butir yang tidak valid
    for y, val, ket in zip(
        posisi, nilai, data_plot["Keterangan"]
    ):
        if ket == "Tidak valid":
            if val >= 0:
                ax.annotate(
                    f"{val:.3f}  Tidak valid",
                    xy=(val, y),
                    xytext=(7, 0),
                    textcoords="offset points",
                    va="center",
                    ha="left",
                    fontsize=9,
                    color="#222222"
                )
            else:
                ax.annotate(
                    f"{val:.3f}  Tidak valid",
                    xy=(val, y),
                    xytext=(-7, 0),
                    textcoords="offset points",
                    va="center",
                    ha="right",
                    fontsize=9,
                    color="#222222"
                )

    # Judul dimensi
    ax.set_title(
        nama_dimensi,
        loc="left",
        fontsize=12,
        fontweight="bold",
        pad=8
    )

    # Grid dan tampilan grafik
    ax.grid(
        axis="x",
        color="#E5E5E5",
        linewidth=0.7,
        zorder=0
    )

    ax.set_axisbelow(True)

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_color("#BDBDBD")
    ax.spines["bottom"].set_color("#BDBDBD")

    ax.tick_params(axis="x", labelsize=9)

    # Keterangan r tabel
    ax.text(
        r_tabel,
        1.02,
        f"r tabel = {r_tabel:.3f}",
        transform=ax.get_xaxis_transform(),
        ha="left",
        va="bottom",
        fontsize=9,
        color="#555555"
    )

# Label sumbu X
axes[-1].set_xlabel("Nilai r hitung", fontsize=10)

# Judul utama
fig.suptitle(
    "Visualisasi Hasil Uji Validitas Butir",
    fontsize=14,
    fontweight="bold",
    y=1.005
)

fig.tight_layout()
plt.show()
```
